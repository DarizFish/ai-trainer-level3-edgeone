```python
# JupyterLite 环境准备：安装本题依赖，并按需下载本题素材
from pathlib import Path
import hashlib
import piplite
from pyodide.http import pyfetch

PACKAGES = ["pandas", "numpy", "matplotlib"]
ASSETS = [{"name": "vehicle_traffic_data.csv", "url": "/labs/1.1.5/vehicle_traffic_data.csv", "size": 52213, "sha256": "8184e508e9787647d82963e0393db26750051d9739afbd2b870faa6674453f86"}]

if PACKAGES:
    await piplite.install(PACKAGES)

for asset in ASSETS:
    target = Path(asset["name"])
    target.parent.mkdir(parents=True, exist_ok=True)
    if target.exists() and hashlib.sha256(target.read_bytes()).hexdigest() == asset["sha256"]:
        continue
    response = await pyfetch(asset["url"])
    payload = await response.bytes()
    digest = hashlib.sha256(payload).hexdigest()
    if digest != asset["sha256"]:
        raise RuntimeError(f"素材校验失败：{asset['name']}")
    target.write_bytes(payload)

print("✓ 1.1.5 环境已就绪", f"依赖 {len(PACKAGES)} 项，素材 {len(ASSETS)} 个")
```
